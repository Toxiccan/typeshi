```python
import numpy as np

x = np.array([1, 2, 3, 4])
y = np.array([2, 4, 6, 8])

n = len(x)

w = 0.0
b = 0.0

learning_rate = 0.1
epochs = 1000
```

```python
for epoch in range(epochs):
     y_hat = w*x + b
     error = y - y_hat

     dw = (-2/n)*np.sum(x * error)
     db = (-2/n)*np.sum(error)

     w = w - learning_rate * dw
     b = b - learning_rate * db

     loss = (1/n) * np.sum(error**2)

     if epoch % 100 == 0:
        print(f"Epoch {epoch},Loss: {loss:.4f}")
```
